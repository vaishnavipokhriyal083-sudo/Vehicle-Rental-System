```python
from abc import ABC, abstractmethod

#--------------------------------------------
#base class  Vehicles
#--------------------------------------------

class Vehicle(ABC):

  def __init__(self, vehicle_id, brand, model, rent_per_day):
   self.__vehicle_id = vehicle_id
   self.__brand = brand
   self.__model = model
   self.__rent_per_day = rent_per_day
   self.__available = True


#Encapsulation -- Geter method
  def get_id(self):
    return self.__vehicle_id

  def get_name(self):
    return self.__brand + " " + self.__model

  def get_rent_per_day(self):
    return self.__rent_per_day

  def is_available(self):
    return self.__available

  def set_available(self,status):
    self.__available = status

  @abstractmethod
  def calculate_rent(self, days):
    pass

  def display(self):
    status = "Available" if self.__available else "Rented"

    print(
        f" ID: {self.__vehicle_id} "
        f" Name: {self.get_name()} "
        f" Rent: ${self.__rent_per_day}/day | "
        f" Status: {status} "
    )

#-----------
#Car Class
#-----------

class Car(Vehicle):

#polymorphism
  def calculate_rent(self, days):
    return days*self.get_rent_per_day()

#-------------
# Bike Class
#-------------

class Bike(Vehicle):
#polymorphism
  def calculate_rent(self, days):
    return days*self.get_rent_per_day()

#-----------
# SUV Class
#-----------

class SUV(Vehicle):
#polymorphism
  def calculate_rent(self, days):
    return days*self.get_rent_per_day()

#-----------------
# Customer Class
#-----------------

class Customer:
  def __init__(self, customer_id, name, phone):
    self.customer_id = customer_id
    self.name = name
    self.phone = phone

  def display(self):
    print(
        f"Customer_id: {self.customer_id}\n"
        f"Name: {self.name}\n"
        f"Phone: {self.phone}\n"
    )

#-----------------------------
# Rental System Class
#-----------------------------

class RentalSystem:
  def __init__(self):
    self.vehicles= []
    self.customers= []


#Add vehicle
  def add_vehicle(self,vehicle):
    self.vehicles.append(vehicle)

#Add customer
  def add_customer(self,customer):
    self.customers.append(customer)

#Dispaly vehicles
  def show_vehicles(self):
     print("\n=====================================VEHICLES====================================\n")

     if not self.vehicles:
      print("No vehicles registered.")
      return

     for vehicle in self.vehicles:
      vehicle.display()


#Display customers
  def show_customers(self):
    print("\n======================================CUSTOMERS===================================\n")

    if not self.customers:
      print("No customers registered.")
      return

    for customer in self.customers:
      customer.display()



#Find Vehicle
  def find_vehicle(self,vehicle_id):

    for vehicle in self.vehicles:

     if vehicle.get_id() == vehicle_id:
      return vehicle

    return None


#Rent Vehicle
  def rent_vehicle(self):

    try:

      vehicle_id=int(input("\n Enter Vehicle ID: "))

      vehicle = self.find_vehicle(vehicle_id)

      if vehicle is None:
       print('Vehicle not found. ')
       return

      if not vehicle.is_available():
        print("Vehicle is already rented.")
        return

      name = input("Enter customer name: \n")
      phone = input("Enter phone number: \n")
      days = int(input("Enter number of days: \n"))

      if days <= 0:
        print("Days must be greater than 0. ")
        return

#Generate Customer ID
      customer_id = len(self.customers)+1

#Create Customer object
      customer = Customer(
       customer_id,
       name,
       phone
    )

      self.add_customer(customer)

#Polymorphism
      total_rent = vehicle.calculate_rent(days)

#Make vehicle unavailable
      vehicle.set_available(False)

      print("\n=============================================Rent========================================================")
      print("Customer ID: ",customer_id )
      print("Customer: ",name )
      print("Phone :", phone)
      print("Vehicle: ", vehicle.get_name() )
      print("Days: ",days )
      print("Rent/Day    : ₹", vehicle.get_rent_per_day())
      print("Rent/Day:$ ",total_rent )
      print("=" * 100)

    except ValueError:
     print("Please enter valid input.")


#return vehicle
  def return_vehicle(self):

   try:
    vehicle_id = int(input("\n Enter Vehicle ID: "))
    vehicle = self.find_vehicle(vehicle_id)

    if vehicle is None:
      print("Vehicle is not found:")
      return

    if vehicle.is_available():
      print("The vehicle is already available.")
      return

    vehicle.set_available(True)

    print("\n Vehicle returned sucessfully!")
    print("Vehicle:", vehicle.get_name())

   except ValueError:
      print("Please enter a valid Vehicle ID.")


#===============================================================================
#Main Program
#===============================================================================

system = RentalSystem()

#===============================================================================
#Adding Vehicles
#===============================================================================

system.add_vehicle(
    Car(101,"Toyota","Camry",1500)
)
system.add_vehicle(
    Bike(102,"Royal Enfield","Classic 350",700)
)
system.add_vehicle(
    SUV(103,"Mahindra","XUV700",2500)
)


#===============================================================================
# Menu
#===============================================================================

while True:
  print("\n")
  print("=======================================================================")
  print("VEHICLE RENTAL SYSTEM")
  print("=======================================================================")


  print("1. Show Vehicles.")
  print("2. Rent Vehicle.")
  print("3. Return Vehicle.")
  print("4. Show Customers.")
  print("5.Exit")

  choice = input("Enter your choice: ")

  if choice == "1":
    system.show_vehicles()

  elif choice == "2":
    system.rent_vehicle()

  elif choice == "3":
    system.return_vehicle()

  elif choice == "4":
    system.show_customers()

  elif choice == '5':
    print("\nThank you for using the system!")
    break

  else:
    print("Invalid choice. Please try again.")
```
